## Loan Eligibility System with New Propositions

This notebook demonstrates a loan eligibility system using propositional logic. We introduce two new propositions:

*   **A**: Customer's age is within the acceptable range.
*   **S**: Customer has a stable employment history.

We will also assume two common propositions from a typical loan eligibility system:

*   **C**: Customer has a good credit score.
*   **I**: Customer has sufficient income.

The new logical rule for eligibility (E) is defined as: `(C ^ I ^ A ^ S) -> E`.

```python
import pandas as pd
from itertools import product

# Define the propositions and the eligibility function
def is_eligible(C, I, A, S):
    """
    Evaluates loan eligibility based on the rule (C ^ I ^ A ^ S) -> E.
    C: Good Credit Score (True/False)
    I: Sufficient Income (True/False)
    A: Acceptable Age Range (True/False)
    S: Stable Employment History (True/False)
    """
    # The rule (C ^ I ^ A ^ S) -> E means that if all conditions are true, then the customer is eligible.
    # If any condition is false, then the conjunction (C ^ I ^ A ^ S) is false, and the implication is still true (Eligible).
    # However, the common interpretation of such a rule in DSS is that eligibility *requires* all conditions to be met.
    # So, E is True if and only if (C AND I AND A AND S) is True.
    return C and I and A and S
```

### Generate Truth Table for the New Logical Rule

Now, let's generate the truth table for the rule `(C ^ I ^ A ^ S) -> E`.

```python
# Generate all possible combinations of truth values for C, I, A, S
propositions = ['C', 'I', 'A', 'S']
truth_values = [True, False]

# Create all permutations of truth values
truth_table_data = []
for c_val, i_val, a_val, s_val in product(truth_values, repeat=len(propositions)):
    eligibility = is_eligible(c_val, i_val, a_val, s_val)
    truth_table_data.append({'C': c_val, 'I': i_val, 'A': a_val, 'S': s_val, 'E': eligibility})

# Create a DataFrame for the truth table
truth_table_df = pd.DataFrame(truth_table_data)

print("Truth Table:")
display(truth_table_df)
```

### Test at least 10 Customer Scenarios

Let's simulate various customer profiles and apply our loan eligibility logic.

```python
customer_scenarios = [
    # Scenario 1: All conditions met (Eligible)
    {'C': True, 'I': True, 'A': True, 'S': True, 'Description': 'Ideal Customer'},
    # Scenario 2: Only credit is bad (Not Eligible)
    {'C': False, 'I': True, 'A': True, 'S': True, 'Description': 'Bad Credit'},
    # Scenario 3: Only income is insufficient (Not Eligible)
    {'C': True, 'I': False, 'A': True, 'S': True, 'Description': 'Insufficient Income'},
    # Scenario 4: Only age is out of range (Not Eligible)
    {'C': True, 'I': True, 'A': False, 'S': True, 'Description': 'Age out of range'},
    # Scenario 5: Only employment is unstable (Not Eligible)
    {'C': True, 'I': True, 'A': True, 'S': False, 'Description': 'Unstable Employment'},
    # Scenario 6: All conditions are false (Not Eligible)
    {'C': False, 'I': False, 'A': False, 'S': False, 'Description': 'All bad'},
    # Scenario 7: Mixed conditions (Not Eligible)
    {'C': True, 'I': False, 'A': True, 'S': False, 'Description': 'Mixed bad'},
    # Scenario 8: Almost ideal, just one false (Not Eligible)
    {'C': True, 'I': True, 'A': False, 'S': True, 'Description': 'Almost ideal, age issue'},
    # Scenario 9: Another mixed condition (Not Eligible)
    {'C': False, 'I': True, 'A': True, 'S': False, 'Description': 'Mixed bad 2'},
    # Scenario 10: Mixed condition with two true, two false (Not Eligible)
    {'C': True, 'I': False, 'A': False, 'S': True, 'Description': 'Mixed bad 3'}
]

# Apply the eligibility function to each scenario
results = []
for scenario in customer_scenarios:
    C_val = scenario['C']
    I_val = scenario['I']
    A_val = scenario['A']
    S_val = scenario['S']
    eligibility = is_eligible(C_val, I_val, A_val, S_val)
    results.append({**scenario, 'E': eligibility})

results_df = pd.DataFrame(results)

print("Customer Scenarios and Eligibility:")
display(results_df)
```

### Identifying Critical Conditions for Loan Eligibility

For the rule `(C ^ I ^ A ^ S) -> E`, all conditions (C, I, A, S) are equally critical. Eligibility `E` is `True` only if ALL conditions are `True` due to the logical AND (`^`) operation. The absence of any single positive condition makes a customer ineligible; thus, no single condition has a 'greater' effect than others.

### Propositional Logic in Decision Support Systems (DSS)

Propositional logic serves as a rule-based component in DSS due to its ability to:

*   **Formally represent rules:** Defines decision criteria using propositions and logical connectives (AND, OR, NOT, IMPLIES).
*   **Provide clear, unambiguous logic:** Ensures consistent rule application, as propositions are either true or false.
*   **Enable automated reasoning:** Allows the DSS to infer conclusions from facts and rules (e.g., loan approval).
*   **Ensure auditability and transparency:** Decisions are traceable to specific rules, offering clear justifications.
*   **Support modularity:** Rules can be managed independently for easier maintenance and adaptation.

**Example:** In our loan system, `(C ^ I ^ A ^ S) -> E` precisely defines eligibility based on customer data, making the decision process transparent and automated.
