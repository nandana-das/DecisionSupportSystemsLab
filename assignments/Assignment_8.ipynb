```python
# Exercise 1: Modified Probability Analysis

# Path durations
success_path = 25
rework_path = 29

# Modified probabilities
p_success = 0.70
p_rework = 0.30

# Calculate expected project duration
expected_duration = (p_success * success_path) + (p_rework * rework_path)

print("EXERCISE 1: MODIFIED PROBABILITY ANALYSIS")
print("-" * 45)
print("Probability of successful evaluation:", p_success)
print("Probability of prototype rework:", p_rework)
print("Success path duration:", success_path, "days")
print("Rework path duration:", rework_path, "days")
print("New Expected Project Duration:", expected_duration, "days")
```

```python
from graphviz import Digraph

dot = Digraph("Exercise 3 GERT Network")
dot.attr(rankdir="LR", size="14")

# Nodes
dot.node("S", "Start", shape="circle")
dot.node("R", "Requirements")
dot.node("D", "Design")
dot.node("P", "Prototype")
dot.node("MR", "Minor Rework")
dot.node("MAR", "Major Rework")
dot.node("I", "Integration")
dot.node("T", "Testing")
dot.node("DEP", "Deployment", shape="doublecircle")

# Main activities
dot.edge("S", "R", label="Requirements\n4 days")
dot.edge("R", "D", label="Design\n5 days")
dot.edge("D", "P", label="Prototype\n6 days")

# Three probabilistic outcomes
dot.edge("P", "I", label="Successful (0.70)\n3 days")
dot.edge("P", "MR", label="Minor Rework (0.20)")
dot.edge("P", "MAR", label="Major Rework (0.10)")

# Merge paths
dot.edge("MR", "I", label="Re-evaluation")
dot.edge("MAR", "I", label="Re-evaluation")

# Final flow
dot.edge("I", "T", label="Testing\n5 days")
dot.edge("T", "DEP", label="Deployment\n2 days")

dot
```

```python
# Exercise 3: Three Outcome GERT Analysis

# Path durations (in days)
success_path = 25
minor_rework_path = 27
major_rework_path = 30

# Probabilities
p_success = 0.70
p_minor = 0.20
p_major = 0.10

# Calculate expected project duration
expected_duration = (
    p_success * success_path +
    p_minor * minor_rework_path +
    p_major * major_rework_path
)

print("EXERCISE 3: THREE-OUTCOME GERT ANALYSIS")
print("-" * 45)
print("Successful Path (0.70):", success_path, "days")
print("Minor Rework Path (0.20):", minor_rework_path, "days")
print("Major Rework Path (0.10):", major_rework_path, "days")
print("-" * 45)
print("Expected Project Duration:", round(expected_duration, 1), "days")
```
