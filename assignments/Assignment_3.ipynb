### Steps 5 & 6: Implement and Test Both Versions

Now that we have both the `original_loan_dt` and `modified_loan_dt`, the next step is to implement a function that can process customer records against these decision tables and then test them with at least 10 customer records.

First, let's define a function that takes a customer's details and a decision table, and returns the appropriate action. We'll need a way to match customer data to the rules in the table, considering 'Don't-Care' conditions.

```python
def make_decision(customer_data, decision_table):
    """
    Evaluates customer data against a decision table to determine an action.
    Handles 'Don't-Care' conditions represented by '-'.

    Args:
        customer_data (dict): A dictionary where keys are conditions (columns of the DT)
                              and values are the customer's specific attributes.
        decision_table (pd.DataFrame): The decision table (original or modified).

    Returns:
        str: The action determined by the decision table, or 'No Rule Matched' if no rule applies.
    """
    for index, rule in decision_table.iterrows():
        match = True
        for condition in customer_data.keys():
            if condition in rule.index:
                rule_value = str(rule[condition]).strip()
                customer_value = str(customer_data[condition]).strip()
                if rule_value != '-' and rule_value != customer_value:
                    match = False
                    break
        if match:
            return rule['Action']
    return "No Rule Matched"

print("Defined the 'make_decision' function.")
```
