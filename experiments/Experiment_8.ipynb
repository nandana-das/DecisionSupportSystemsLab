```python
from graphviz import Digraph

# Create GERT network
dot = Digraph("GERT Network")
dot.attr(rankdir="LR", size="12")

# Nodes
dot.node("S", "Start", shape="circle")
dot.node("R", "Requirements")
dot.node("D", "Design")
dot.node("P", "Prototype")
dot.node("E", "Evaluation")
dot.node("RW", "Rework")
dot.node("I", "Integration")
dot.node("T", "Testing")
dot.node("DEP", "Deployment", shape="doublecircle")

# Main flow
dot.edge("S", "R", label="Requirements\n4 days")
dot.edge("R", "D", label="Design\n5 days")
dot.edge("D", "P", label="Prototype\n6 days")

# Probabilistic paths
dot.edge("P", "E", label="Evaluate")
dot.edge("E", "I", label="Success (0.8)\n3 days")
dot.edge("E", "RW", label="Rework (0.2)\n4 days")
dot.edge("RW", "I", label="Re-evaluation")

# Final flow
dot.edge("I", "T", label="Integration\n5 days")
dot.edge("T", "DEP", label="Testing & Deployment\n2 days")

# Display
dot
```

```python
# Path durations
success_path = 25
rework_path = 29

# Probabilities
p_success = 0.8
p_rework = 0.2

# Expected project duration
expected_duration = (p_success * success_path) + (p_rework * rework_path)

print("Success Path Duration:", success_path, "days")
print("Rework Path Duration:", rework_path, "days")
print("Expected Project Duration:", expected_duration, "days")
```

```python
print("GERT NETWORK EVALUATION")
print("-" * 35)
print("Probability of successful prototype:", p_success)
print("Probability of prototype rework:", p_rework)
print("Duration through success path:", success_path, "days")
print("Duration through rework path:", rework_path, "days")
print("Expected project duration:", expected_duration, "days")

print("\nInterpretation:")
print("The project has an 80% probability of proceeding successfully.")
print("There is a 20% probability that prototype rework will be required.")
print("The probabilistic rework path increases the expected project duration.")
```
