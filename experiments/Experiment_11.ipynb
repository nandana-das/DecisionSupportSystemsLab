# Experiment No. 11: Basic Intelligent Decision Support System Using Machine Learning
**Aim:** To implement a basic Intelligent Decision Support System (IDSS) that predicts whether a customer is likely to be approved or rejected for a loan, evaluates the model, and provides a decision-support recommendation.

**Note:** This is an educational demonstration using a very small illustrative dataset. It must not be used to make real lending decisions.

## 1. Import libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 2. Create the historical customer dataset

```python
data = {
    "Age": [25, 32, 45, 29, 41, 52, 27, 36, 48, 30, 39, 26, 43, 34, 50],
    "Income": [35000, 60000, 85000, 42000, 75000, 90000, 38000, 65000, 70000, 45000, 58000, 32000, 82000, 55000, 95000],
    "Credit_Score": [620, 720, 780, 650, 740, 800, 610, 710, 730, 640, 690, 600, 760, 680, 810],
    "Loan_Amount": [15000, 20000, 30000, 18000, 25000, 40000, 12000, 22000, 28000, 16000, 20000, 10000, 30000, 18000, 45000],
    "Debt": [8000, 5000, 4000, 12000, 7000, 6000, 9000, 6000, 8000, 10000, 7000, 7000, 5000, 9000, 5000],
    "Employment": ["Salaried", "Salaried", "Salaried", "Self-Employed", "Salaried", "Self-Employed", "Salaried", "Salaried", "Self-Employed", "Self-Employed", "Salaried", "Salaried", "Salaried", "Self-Employed", "Salaried"],
    "Loan_Decision": ["Rejected", "Approved", "Approved", "Rejected", "Approved", "Approved", "Rejected", "Approved", "Approved", "Rejected", "Approved", "Rejected", "Approved", "Rejected", "Approved"]
}
df = pd.DataFrame(data)
print("INTELLIGENT DSS - LOAN DECISION")
display(df)
```

## 3. Inspect the dataset for missing values and data types

```python
print("Missing values per column:")
display(df.isnull().sum().to_frame("Missing Values"))
print("\nData types:")
print(df.dtypes)
```

## 4. Encode categorical variables

```python
df["Employment"] = df["Employment"].map({"Salaried": 0, "Self-Employed": 1})
df["Loan_Decision"] = df["Loan_Decision"].map({"Rejected": 0, "Approved": 1})
display(df.head())
```

## 5. Separate features and target

```python
feature_columns = ["Age", "Income", "Credit_Score", "Loan_Amount", "Debt", "Employment"]
X = df[feature_columns]
y = df["Loan_Decision"]
print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

## 6. Split data into training and testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 7. Create and train the Decision Tree Classifier

```python
model = DecisionTreeClassifier(criterion="entropy", max_depth=3, random_state=42)
model.fit(X_train, y_train)
print("Decision Tree model trained successfully.")
```

## 8. Predict test-set outcomes

```python
y_pred = model.predict(X_test)
results = pd.DataFrame({"Actual": y_test.map({0: "Rejected", 1: "Approved"}),
                        "Predicted": pd.Series(y_pred, index=y_test.index).map({0: "Rejected", 1: "Approved"})})
display(results)
```

## 9. Evaluate the model

```python
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")
print("\nConfusion matrix (rows = actual, columns = predicted; order: Rejected, Approved):")
print(confusion_matrix(y_test, y_pred, labels=[0, 1]))
print("\nClassification report:")
print(classification_report(y_test, y_pred, labels=[0, 1], target_names=["Rejected", "Approved"], zero_division=0))
```

## 10. Enter details for a new customer

```python
# Example customer from the experiment sheet
new_customer = pd.DataFrame({
    "Age": [38],
    "Income": [70000],
    "Credit_Score": [735],
    "Loan_Amount": [25000],
    "Debt": [6000],
    "Employment": [0]  # 0 = Salaried, 1 = Self-Employed
})
display(new_customer)
```

## 11. Predict the new customer's decision

```python
prediction = int(model.predict(new_customer)[0])
probability = model.predict_proba(new_customer)[0]
decision_map = {0: "REJECTED", 1: "APPROVED"}
decision = decision_map[prediction]
print("Predicted decision:", decision)
print(f"Probability of rejection: {probability[0] * 100:.2f}%")
print(f"Probability of approval: {probability[1] * 100:.2f}%")
```

## 12. Generate a decision-support recommendation

```python
if prediction == 1:
    print("Recommendation: The customer may be considered for loan approval based on the trained model.")
else:
    print("Recommendation: The customer may require additional assessment before loan approval.")
print("\nNote: This is an educational DSS demonstration and should not be used as a real lending decision.")
```

## Result
A basic Intelligent Decision Support System using a Decision Tree classifier was implemented with Pandas and Scikit-learn. The notebook checks the data, encodes categorical variables, trains and evaluates the classifier, predicts a decision for a new customer, and prints a decision-support recommendation.
