# Experiment 2: Automated Loan Approval System with Decision Table

## Algorithm

**Input:**
Customer Information: Credit Score, Monthly Income, Employment Status, Existing Loan, Customer Age, Debt-to-Income Ratio

**Output:**
Loan Decision

**Steps**
1.  Start
2.  Read customer details.
3.  Check if Customer Age is within the acceptable range (e.g., 25-60).
4.  Check Debt-to-Income Ratio (DTI).
5.  Check if Credit Score ≥ 750.
6.  Check if Monthly Income ≥ ₹50,000.
7.  Check Employment Status.
8.  Check Existing Loan.
9.  Apply Decision Table rules (considering Age and DTI first).
10. Display the loan decision.
11. Stop.

```python
import pandas as pd

# Enhanced Function for Loan Decision with Customer Age and Debt-to-Income Ratio
def loan_decision_enhanced(credit_score, income, employed, existing_loan, customer_age, dti_ratio):

    # Rule for Customer Age
    if customer_age < 25 or customer_age > 60:
        return "Manual Review" # Age outside ideal range

    # Rule for Debt-to-Income Ratio
    if dti_ratio > 0.40:
        return "Reject Loan" # High DTI
    elif 0.30 <= dti_ratio <= 0.40:
        return "Manual Review" # Borderline DTI

    # Original rules, now applied after age and DTI checks
    if credit_score >= 750 and income >= 50000 and employed == "Yes" and existing_loan == "No":
        return "Approve Loan"
    elif credit_score >= 750 and employed == "Yes":
        return "Manual Review"
    elif income >= 50000 and employed == "Yes":
        return "Manual Review"
    elif credit_score >= 750 and income >= 50000:
        return "Manual Review"
    else:
        return "Reject Loan"
```

## Decision Rules (Implemented in `loan_decision_enhanced`)

```python
# Sample Customer Records as provided in the PDF (original)
original_customers = pd.DataFrame({
    "Customer":[1,2,3,4,5],
    "Credit Score":[780,760,690,720,810],
    "Income":[65000,45000,70000,30000,90000],
    "Employed":["Yes","Yes","Yes","No","No"],
    "Existing Loan":["No","Yes","No","No","Yes"]
})

print("\nOriginal Customer Details and Loan Decisions:\n")
original_customers["Decision"] = original_customers.apply(
    lambda row:
    loan_decision_enhanced( # Using enhanced function but with dummy values for new params
        row["Credit Score"],
        row["Income"],
        row["Employed"],
        row["Existing Loan"],
        customer_age=35, # Dummy age for comparison with original logic
        dti_ratio=0.25   # Dummy DTI for comparison with original logic
    ),
    axis=1
)
display(original_customers)

# New, expanded sample customer records (at least 10) with Age and DTI
enhanced_customers = pd.DataFrame({
    "Customer": range(1, 12),
    "Credit Score": [780, 760, 690, 720, 810, 650, 790, 700, 820, 580, 750],
    "Income": [65000, 45000, 70000, 30000, 90000, 40000, 80000, 55000, 100000, 25000, 60000],
    "Employed": ["Yes", "Yes", "Yes", "No", "No", "Yes", "Yes", "Yes", "No", "Yes", "Yes"],
    "Existing Loan": ["No", "Yes", "No", "No", "Yes", "No", "No", "Yes", "No", "Yes", "No"],
    "Customer Age": [35, 22, 58, 65, 40, 28, 50, 62, 30, 20, 45],
    "Debt-to-Income Ratio": [0.25, 0.35, 0.15, 0.45, 0.20, 0.30, 0.22, 0.38, 0.10, 0.50, 0.28]
})

print("\nEnhanced Customer Details and Loan Decisions (with Age and DTI):\n")
enhanced_customers["Decision"] = enhanced_customers.apply(
    lambda row:
    loan_decision_enhanced(
        row["Credit Score"],
        row["Income"],
        row["Employed"],
        row["Existing Loan"],
        row["Customer Age"],
        row["Debt-to-Income Ratio"]
    ),
    axis=1
)
display(enhanced_customers)
```
