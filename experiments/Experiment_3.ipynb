## Basic Decision Table for Insurance Claims

First, we define the basic decision table for insurance claims based on the conditions and decisions outlined in the problem statement. This table serves as our initial reference.

```python
# ------------------------------------------------------------
# Decision Table - Version 1
# Basic Decision Table
# ------------------------------------------------------------
decision_table = pd.DataFrame({
    "Policy_Active": ["Yes", "Yes", "Yes", "No"],
    "Claim_Covered": ["Yes", "Yes", "No", "-"],
    "Documents_Complete": ["Yes", "No", "Yes", "-"],
    "Within_Limit": ["Yes", "Yes", "Yes", "-"],
    "Decision": [
        "Approve Claim",
        "Manual Review",
        "Reject Claim",
        "Reject Claim"
    ]
})

print("\n=============================================")
print("BASIC DECISION TABLE")
print("=============================================")
display(decision_table)
```

## Function to Simulate Insurance Decisions

Next, we implement a Python function that takes a customer's claim details as input and applies the rules defined in our decision table to determine the appropriate decision (Approve, Manual Review, or Reject).

```python
# ------------------------------------------------------------
# Function to simulate the Decision Table
# ------------------------------------------------------------
def insurance_decision(policy_active, claim_covered, documents_complete, within_limit):
    # Rule 1: All positive conditions for approval
    if (policy_active == "Yes" and
            claim_covered == "Yes" and
            documents_complete == "Yes" and
            within_limit == "Yes"):
        return "Approve Claim"

    # Rule 2: Policy active, covered, but documents incomplete for manual review
    elif (policy_active == "Yes" and
          claim_covered == "Yes" and
          documents_complete == "No"):
        return "Manual Review"

    # Rule 3: Policy active but claim not covered for rejection
    elif (policy_active == "Yes" and
          claim_covered == "No"):
        return "Reject Claim"

    # Rule 4: Policy not active leads to rejection (Don't-Care conditions on others)
    elif policy_active == "No":
        return "Reject Claim"

    # Other cases default to manual review if no specific rule matches
    else:
        return "Manual Review"
```

## Test Cases and Simulation Results

We define a set of sample customer cases with varying claim conditions. We then apply the `insurance_decision` function to each case to simulate the decision-making process and display the results, including a summary of the decisions.

```python
# ------------------------------------------------------------
# Test Cases
# ------------------------------------------------------------
customers = pd.DataFrame({
    "Customer_ID": [
        "C001", "C002", "C003", "C004", "C005", "C006"
    ],
    "Policy_Active": [
        "Yes", "Yes", "Yes", "No", "Yes", "Yes"
    ],
    "Claim_Covered": [
        "Yes", "Yes", "No", "Yes", "Yes", "No"
    ],
    "Documents_Complete": [
        "Yes", "No", "Yes", "No", "Yes", "Yes"
    ],
    "Within_Limit": [
        "Yes", "Yes", "Yes", "Yes", "No", "Yes"
    ]
})

# ------------------------------------------------------------
# Apply Decision Table
# ------------------------------------------------------------
customers["Decision"] = customers.apply(
    lambda row: insurance_decision(
        row["Policy_Active"],
        row["Claim_Covered"],
        row["Documents_Complete"],
        row["Within_Limit"]
    ),
    axis=1
)

# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------
print("\n=============================================")
print("SIMULATION RESULTS")
print("=============================================")
display(customers)

# ------------------------------------------------------------
# Decision Summary
# ------------------------------------------------------------
print("\n=============================================")
print("DECISION SUMMARY")
print("=============================================")
summary = customers["Decision"].value_counts()
print(summary)
```
